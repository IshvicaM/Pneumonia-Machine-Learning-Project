```python
import tensorflow as tf # Import TensorFlow in order to have machine learning framework for image classification.
mnist = tf.keras.datasets.mnist # 28 x 28 images of handwritten digits 0-9
(x_train, y_train), (x_test, y_test) = mnist.load_data() # Train and test dataset.
x_train = tf.keras.utils.normalize(x_train, axis=1) # Review what axis=1 means again..
x_test = tf.keras.utils.normalize(x_test, axis=1) # But for x_test.

model = tf.keras.models.Sequential() # Layer-by-layer stack, simplest neural network architecture, one input tensor, one output tensor.
model.add(tf.keras.layers.Flatten()) # Flattens the model as Tensor only processes one dimension [1D array of images].
model.add(tf.keras.layers.Dense(128, activation='relu')) # Activation layer, determines if signal should move forward, outputs input directly if positive, returns zero.
model.add(tf.keras.layers.Dense(128, activation='relu')) # Another activation layer, mathematical function max(0, x).
model.add(tf.keras.layers.Dense(10, activation='softmax')) #output layer (not 128, classifcation is number of classifactions, activation function is probaility distrubition)

model.compile(optimizer='adam', loss = 'sparse_categorical_crossentropy', metrics=['accuracy']) # Employed use of the "adam" optimizer, defined our loss function, and tracked overall accuracy.
model.fit(x_train, y_train, epochs=3) # This basically means the model will do three rounds of training.
```

```python
val_loss, val_acc = model.evaluate(x_test, y_test) # testing performance of machine learning model
print(val_loss, val_acc)
```

```python
model.save('epic_numb_reader.keras') # saves model into a single zip folder
```

```python
new_model = tf.keras.models.load_model('epic_numb_reader.keras') # makes testing model = to trainng model that was saved above.
```
